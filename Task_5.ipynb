````python
from transformers import pipeline

print("Initializing Support Intelligence...")

support_engine = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
print("System online. Type 'quit' to terminate session.\n")

def craft_reply(classification, confidence):
    status = classification.upper()


    if status == "POSITIVE":
        return "Awesome! We are thrilled to hear that. Is there anything else we can do for you today?"
    elif status == "NEGATIVE":
        if confidence >= 0.85:
            return "I sincerely apologize for this unacceptable experience. I am escalating your case to a senior manager right now."
        else:
            return "I am sorry to hear you're experiencing difficulties. Let's work together to fix this immediately."
    else:
        return "Thank you for the update. How would you like to proceed?"

def start_interaction():
    print("Virtual Agent: Hello! Welcome to our helpdesk. What can I assist you with today?")

    while True:
        try:
            customer_message = input("\nCustomer: ")
        except EOFError:
            break

        if customer_message.strip().lower() == 'quit':
            print("Virtual Agent: Thank you for reaching out. Have a great day!")
            break

        if not customer_message.strip():
            print("Virtual Agent: Please enter a valid query.")
            continue


        analysis_result = support_engine(customer_message)[0]
        sentiment_label = analysis_result['label']
        confidence_metric = round(analysis_result['score'], 4)


        agent_response = craft_reply(sentiment_label, confidence_metric)


        print(f"\n[Telemetry -> Classification: {sentiment_label} | Reliability Index: {confidence_metric}]")
        print(f"Virtual Agent: {agent_response}")

start_interaction()
````

````python
import pandas as pd


data = {
    "customer_message": [
        "I love this product, it works amazingly well!",
        "Your service is absolute garbage and nothing works.",
        "Can you please tell me what my account balance is?",
        "Thank you so much for the quick help, you're the best!",
        "This is the worst experience I have ever had with a company.",
        "Could you provide instructions on how to reset my password?",
        "Absolute waste of money. I want a refund immediately.",
        "The app is okay, but it could use a few improvements.",
        "I am so happy with the customer support team today!",
        "Stop calling me, or I will report your business."
    ],
    "true_sentiment": [
        "POSITIVE", "NEGATIVE", "NEUTRAL",
        "POSITIVE", "NEGATIVE", "NEUTRAL",
        "NEGATIVE", "NEUTRAL", "POSITIVE", "NEGATIVE"
    ]
}


df_dataset = pd.DataFrame(data)


file_path = "customer_sentiment_dataset.csv"
df_dataset.to_csv(file_path, index=False)

print(f"Dataset successfully created and saved locally as '{file_path}'!")
print("\nDataset Preview:")
print(df_dataset.to_string())
````
